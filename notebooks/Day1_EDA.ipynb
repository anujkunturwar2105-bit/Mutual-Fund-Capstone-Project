```python
import pandas as pd

# Load datasets
fund_master = pd.read_csv("../Data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../Data/raw/02_nav_history.csv")

# Find AMFI codes that are in fund_master but missing in nav_history
missing_codes = fund_master[
    ~fund_master["amfi_code"].isin(nav_history["amfi_code"])
]

print("Number of missing AMFI codes:", len(missing_codes))
display(missing_codes.head())

# Save the report
missing_codes.to_csv("../reports/data_quality_report.csv", index=False)

print("✅ Data quality report saved to reports/data_quality_report.csv")
```
