```python
import pandas as pd
from pathlib import Path

# Project paths
BASE_DIR = Path.cwd()
if not (BASE_DIR / "Data").exists() and (BASE_DIR.parent / "Data").exists():
    BASE_DIR = BASE_DIR.parent
RAW_DIR = BASE_DIR / "Data" / "raw"
PROCESSED_DIR = BASE_DIR / "Data" / "processed"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Files to clean
datasets = {
    
    
    "03_aum_by_fund_house.csv": "clean_aum.csv",
    "04_monthly_sip_inflows.csv": "clean_monthly_sip.csv",
    "05_category_inflows.csv": "clean_category_inflows.csv",
    "06_industry_folio_count.csv": "clean_industry_folio.csv",

    "09_portfolio_holdings.csv": "clean_portfolio_holdings.csv",
    "10_benchmark_indices.csv": "clean_benchmark_indices.csv",
}

print("=" * 70)
print("STARTING DATA CLEANING")
print("=" * 70)

for input_file, output_file in datasets.items():

    print(f"\nProcessing: {input_file}")

    try:
        df = pd.read_csv(RAW_DIR / input_file)

        print(f"Original Shape : {df.shape}")

        # Remove duplicate rows
        duplicates = df.duplicated().sum()
        df = df.drop_duplicates()

        # Remove completely empty rows
        df = df.dropna(how="all")

        # Strip spaces from column names
        df.columns = df.columns.str.strip()

        # Strip spaces from text columns
        for col in df.select_dtypes(include="object").columns:
            df[col] = df[col].astype(str).str.strip()

        # Convert date columns
        for col in df.columns:
            if "date" in col.lower():
                try:
                    df[col] = pd.to_datetime(df[col], errors="coerce")
                except:
                    pass

        # Convert numeric columns where possible
        for col in df.columns:
            if any(word in col.lower() for word in [
                "amount",
                "nav",
                "return",
                "ratio",
                "expense",
                "aum",
                "value",
                "price"
            ]):
                try:
                    df[col] = pd.to_numeric(df[col], errors="coerce")
                except:
                    pass

        # Remove duplicate rows again after conversions
        df = df.drop_duplicates()

        # Save cleaned file
        output_path = PROCESSED_DIR / output_file
        df.to_csv(output_path, index=False)

        print(f"Duplicates Removed : {duplicates}")
        print(f"Final Shape : {df.shape}")
        print(f"Saved : {output_file}")

    except Exception as e:
        print(f"Error processing {input_file}")
        print(e)

print("\n" + "=" * 70)
print("ALL DATASETS CLEANED SUCCESSFULLY")
print("=" * 70)
```
